```python
#Today's topic
MAX() function
Goal: learn how to find the largest value in a numeric or date column using MAX()
```

```python
#Find the highest salary among all employees
pd.read_sql("""
SELECT MAX(salary) AS highest_salary
FROM employees;
""", conn)
```

```python
#Find the maximum product price
pd.read_sql("""
SELECT MAX(price) AS max_price
FROM products;
""", conn)
```

```python
#Find the latest order date in the orders table.
pd.read_sql("""
SELECT MAX(order_date) AS latest_order_date
FROM orders;
""", conn)
```

```python
#Find the highest stock quantity for products in the Electronics category
pd.read_sql("""
SELECT MAX(stock_quantity) AS highest_stock
FROM products
WHERE category = 'Electronics';
""", conn)
```

```python
#Find the highest salary of employees whose department_id is 2
pd.read_sql("""
SELECT MAX(salary) AS highest_salary
FROM employees
WHERE department_id = 2;
""", conn)
```
