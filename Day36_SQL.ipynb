```python
import sqlite3
import pandas as pd
```

```python
conn = sqlite3.connect("/content/drive/MyDrive/SQL/company.db")
```

```python
pd.read_sql("""
SELECT name
FROM sqlite_master
WHERE type='table';
""", conn)
```

```python
#find the lowest product price in the products table
pd.read_sql("""
SELECT MIN(price) AS lowest_price
FROM products;
""", conn)
```

```python
#find the minimum salary among all employees
pd.read_sql("""
SELECT MIN(salary) AS minimum_salary
FROM employees;
""", conn)
```

```python
#find the earliest order_date in the orders table
pd.read_sql("""
SELECT MIN(order_date) AS earliest_order_date
FROM orders;
""", conn)
```

```python
#find the minimum stock quantity for products in the Electronics category
pd.read_sql("""
SELECT MIN(stock_quantity) AS minimum_stock
FROM products
WHERE category = 'Electronics';
""", conn)
```
