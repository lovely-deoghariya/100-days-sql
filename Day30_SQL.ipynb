```python
#Display all customers whose customer_name starts with "A" , sorted alphabetically
pd.read_sql("""
SELECT *
FROM customers
WHERE first_name LIKE 'A%'
ORDER BY first_name;
""", conn)
```

```python
#Display employees whose first_name contains the letter "a" sorted by first_name
pd.read_sql("""
SELECT *
FROM employees
WHERE first_name LIKE '%a%'
ORDER BY first_name;
""", conn)
```

```python
#Display products whose product_name ends with "r" sorted by price in descending order
pd.read_sql("""
SELECT *
FROM products
WHERE product_name LIKE '%r'
ORDER BY price DESC;
""", conn)
```

```python
#Display department whose department_name contains the letter "e" sorted alphabetically
pd.read_sql("""
SELECT *
FROM departments
WHERE department_name LIKE '%e%'
ORDER BY department_name;
""", conn)
```

```python
#Display customers whose city starts with "B" and ends with "e" sorted by customer_name
pd.read_sql("""
SELECT *
FROM customers
WHERE city LIKE 'B%e'
ORDER BY first_name;
""", conn)
```
