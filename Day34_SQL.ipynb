```python
#Today's topic
SUM() function
Goal: learn how to calculate the total value of numeric data using SUM()
```

```python
#find the total quantity of all orders
pd.read_sql("""
SELECT SUM(quantity) AS total_quantity
FROM orders;
""", conn)
```

```python
#find the total stock quantity of all products
pd.read_sql("""
SELECT SUM(stock_quantity) AS total_stock
FROM products;
""", conn)
```

```python
#find the total salary of employees whose department_id is 1
pd.read_sql("""
SELECT SUM(salary) AS total_salary
FROM employees
WHERE department_id = 1;
""", conn)
```

```python
#find the total quantity ordered where the order quantity is greater than 2
pd.read_sql("""
SELECT SUM(quantity) AS total_ordered
FROM orders
WHERE quantity > 2;
""", conn)
```

```python
#find the total price of all products in the electronics category
pd.read_sql("""
SELECT SUM(price) AS total_price
FROM products
WHERE category = 'Electronics';
""", conn)
```
