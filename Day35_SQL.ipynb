```python
#Today's topic
AVG() function
Goal: learn how to calculate the average value of numeric data using AVG()
```

```python
#find the average salary of all employees
pd.read_sql("""
SELECT AVG(salary) AS average_salary
FROM employees;
""", conn)
```

```python
#find the average price of all products
pd.read_sql("""
SELECT AVG(price) AS average_price
FROM products;
""", conn)
```

```python
#find the average quantity ordered in the order table
pd.read_sql("""
SELECT AVG(quantity) AS average_quantity
FROM orders;
""", conn)
```

```python
#find the average salary of employees whose department_id is 2
pd.read_sql("""
SELECT AVG(salary) AS average_salary
FROM employees
WHERE department_id = 2;
""", conn)
```
