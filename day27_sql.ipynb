```python
#Display customers whose customer_name starts with "A" and has exactly one more character
pd.read_sql("""
SELECT *
FROM customers
WHERE first_name LIKE 'A_';
""", conn)
```

```python
#Display employees whose first_name has exactly five characters
pd.read_sql("""
SELECT *
FROM employees
WHERE first_name LIKE '_____';
""", conn)
```

```python
#Display products whose product_name starts with "s" followed by exactly one character and then any remaining characters
pd.read_sql("""
SELECT *
FROM products
WHERE product_name LIKE 'S_%';
""", conn)
```

```python
#Display departments whose department_name has "e" as the second character
pd.read_sql("""
SELECT *
FROM departments
WHERE department_name LIKE '_e%';
""", conn)
```

```python
#find customers whose city starts with "B" has exactly one unknown character and then any remaining character
pd.read_sql("""
SELECT *
FROM customers
WHERE city LIKE 'B_%';
""", conn)
```
