```python
#Today's topic
COUNT() with WHERE
Goal: learn how to count only the rows that match a condition
```

```python
#count the total number of customers
pd.read_sql("""
SELECT COUNT(*)
FROM customers;
""", conn)
```

```python
#count how many products belong to the electronics category
pd.read_sql("""
SELECT COUNT(*)
FROM products
WHERE category = 'Electronics';
""", conn)
```

```python
#count how many employees work in the sales department using department_id
pd.read_sql("""
SELECT COUNT(*)
FROM employees
WHERE department_id = 1;
""", conn)
```

```python
#count the number of orders placed where the quantity is greater than 3
pd.read_sql("""
SELECT COUNT(*)
FROM orders
WHERE quantity > 3;
""", conn)
```

```python
#count how many employees earn a salary greater than 60000
pd.read_sql("""
SELECT COUNT(*)
FROM employees
WHERE salary > 60000;
""", conn)
```
