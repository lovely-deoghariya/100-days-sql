```python
#Today's topic
Mixed practice with LIKE,% and _ wildcard
```

```python
#Display customers whose customer_name starts with the letter "s"
pd.read_sql("""
SELECT *
FROM customers
WHERE first_name LIKE 'S%';
""", conn)
```

```python
#Display employees whose first_name ends with the letter "n"
pd.read_sql("""
SELECT *
FROM employees
WHERE first_name LIKE '%n';
""", conn)
```

```python
#Display products whose product_name contains the letter "pro"
pd.read_sql("""
SELECT *
FROM products
WHERE product_name LIKE '%pro%';
""", conn)
```

```python
#Display department whose department_name has exactly one character before the letter "e"
pd.read_sql("""
SELECT *
FROM departments
WHERE department_name LIKE '_e%';
""", conn)
```

```python
#Display customers whose city starts with "B" and ends with "e"
pd.read_sql("""
SELECT *
FROM customers
WHERE city LIKE 'B%e';
""", conn)
```
