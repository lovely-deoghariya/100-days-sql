```python
#Today's topic
Using LIKE to find values that end with specific letters
```

```python
#Display all customers whose customer_name ends with the letter "n"
pd.read_sql("""
SELECT *
FROM customers
WHERE first_name LIKE '%n';
""", conn)
```

```python
#Display all employees whose first_name ends with the letter "a"
pd.read_sql("""
SELECT *
FROM employees
WHERE first_name LIKE '%a';
""", conn)
```
